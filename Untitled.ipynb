```python
from pypdf import PdfReader

pdf_path = r"C:\Users\Acer\Downloads\5.3 MLBasics-Hyperparam.pdf"

reader = PdfReader(pdf_path)

text = ""

for page in reader.pages:
    text += page.extract_text() or ""

print("Total pages:", len(reader.pages))
print("Total characters:", len(text))
```

```python
chunk_size = 500
overlap = 100

chunks = []

for i in range(0, len(text), chunk_size - overlap):
    chunk = text[i:i + chunk_size]
    chunks.append(chunk)

print("Total chunks:", len(chunks))
```

```python
print("Total chunks:", len(chunks))

print("\nFirst chunk:\n")
print(chunks[0])
```

```python
embeddings = model.encode(chunks)

print("Number of chunks:", len(chunks))
print("Embeddings shape:", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
question = "What are hyperparameters in machine learning?"
question_embedding = model.encode([question])

scores = cosine_similarity(question_embedding, embeddings)[0]
```

```python
print("Similarity scores:", scores)
```

```python
import numpy as np

top_indices = np.argsort(scores)[::-1][:3]

for i in top_indices:
    print(f"\nChunk {i} | Similarity: {scores[i]:.4f}")
    print(chunks[i])
```

```python
from langchain_core.prompts import PromptTemplate

fusion_prompt = PromptTemplate(
    input_variables=["question"],
    template="""Generate exactly 3 different search queries for the question below.
Write only one query per line.
Do not add numbering or explanations.

Original question: {question}
"""
)

question = "Why are hyperparameters used in machine learning?"

query_response = llm.invoke(
    fusion_prompt.format(question=question)
)

queries = [
    line.strip()
    for line in query_response.content.splitlines()
    if line.strip()
]

print("Original question:", question)
print("\nGenerated queries:")
for q in queries:
    print(q)
```

```python
ranked_results = []

for q in queries:
    docs = vectorstore.similarity_search(q, k=3)
    ranked_results.append(docs)

for i, docs in enumerate(ranked_results, start=1):
    print(f"\nSearch {i}:")
    for rank, doc in enumerate(docs, start=1):
        print(f"Rank {rank}: {doc.page_content[:100]}")
```

```python
from collections import defaultdict

rrf_scores = defaultdict(float)

rrf_k = 60

for docs in ranked_results:
    for rank, doc in enumerate(docs, start=1):
        chunk_id = doc.metadata["chunk_id"]

        rrf_scores[chunk_id] += 1 / (rrf_k + rank)

print("RRF scores:")
for chunk_id, score in sorted(
    rrf_scores.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(f"Chunk {chunk_id}: {score:.6f}")
```

```python
final_ranked_chunks = sorted(
    rrf_scores.items(),
    key=lambda item: item[1],
    reverse=True
)

for chunk_id, score in final_ranked_chunks:
    print(f"Chunk ID: {chunk_id}, RRF Score: {score:.6f}")
```

```python
chunk_lookup = {
    doc.metadata["chunk_id"]: doc
    for doc in documents
}

final_docs = [
    chunk_lookup[chunk_id]
    for chunk_id, score in final_ranked_chunks
]

for rank, doc in enumerate(final_docs, start=1):
    print(f"\nFinal Rank: {rank}")
    print(f"Chunk ID: {doc.metadata['chunk_id']}")
    print(doc.page_content[:200])
```

```python
question = "Why are hyperparameters used in machine learning?"

context = "\n\n".join(
    doc.page_content for doc in final_docs[:3]
)

prompt = f"""
Answer the question using only the context below.
If the context does not contain the answer, say you don't know.

Context:
{context}

Question:
{question}

Answer:
"""

response = llm.invoke(prompt)
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

decomposition_prompt = PromptTemplate(
    input_variables=["question"],
    template="""Break the question into 3 smaller questions
that together help answer the original question.

Return only the questions, one per line.

Original question: {question}
"""
)

question = (
    "How do hyperparameters affect model performance, "
    "and how do we choose the best values?"
)

formatted_prompt = decomposition_prompt.format(
    question=question
)

print(formatted_prompt)
```

```python
response = llm.invoke(formatted_prompt)

print(response.content)
```

```python
sub_questions = [
    line.strip()
    for line in response.content.splitlines()
    if line.strip()
]

print(sub_questions)
```

```python
all_results = []

for sub_question in sub_questions:
    docs = vectorstore.similarity_search(sub_question, k=3)
    all_results.extend(docs)
```
